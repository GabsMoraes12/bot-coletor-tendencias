```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
from datetime import datetime

print("🔄 Iniciando o Bot Coletor de Tendências...")

url = "https://news.ycombinator.com/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}

try:
    resposta = requests.get(url, headers=headers)

    if resposta.status_code == 200:
        soup = BeautifulSoup(resposta.text, 'html.parser')

        links = soup.find_all('span', class_='titleline')

        titulos_encontrados = []
        for link in links[:10]:
            texto_titulo = link.text
            titulos_encontrados.append(texto_titulo)

        data_atual = datetime.now().strftime("%d/%m/%Y %H:%M")

        relatorio = {
            "Tópico / Tendência": titulos_encontrados,
            "Data da Coleta": [data_atual] * len(titulos_encontrados),
            "Status": ["Análise Recomendada"] * len(titulos_encontrados)
        }

        df = pd.DataFrame(relatorio)

        nome_arquivo = "Tendencias_Mercado_Coletadas.xlsx"
        df.to_excel(nome_arquivo, index=False)

        print(f"\n🔥 Sucesso! O Bot coletou as 10 principais tendências globais.")
        print(f"📄 Arquivo '{nome_arquivo}' gerado com sucesso para a equipe de marketing!")
        print("\n--- Amostra dos dados coletados ---")
        print(df[["Tópico / Tendência"]].head(3))

    else:
        print(f"❌ Erro ao acessar o site. Status Code: {resposta.status_code}")

except Exception as e:
    print(f"❌ Ocorreu um erro na execução do bot: {e}")
```
